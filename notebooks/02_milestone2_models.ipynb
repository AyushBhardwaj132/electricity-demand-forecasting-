```python
import sys
from pathlib import Path

# Add project root to sys.path
project_root = Path.cwd().parent
if str(project_root) not in sys.path:
    sys.path.append(str(project_root))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import adfuller, kpss
from statsmodels.tsa.seasonal import seasonal_decompose
print("Core libraries loaded successfully.")
```

## 1. Load Preprocessed Dataset & Stationarity Analysis (ADF & KPSS)

```python
data_path = project_root / "data" / "processed" / "cleaned_daily.csv"
df = pd.read_csv(data_path)
df["yyyymmdd"] = pd.to_datetime(df["yyyymmdd"])
df = df.sort_values("yyyymmdd").reset_index(drop=True)
series = df.set_index("yyyymmdd")["India: EnergyMet"]

# Statistical Stationarity: ADF & KPSS tests
adf_raw = adfuller(series.dropna())
kpss_raw = kpss(series.dropna(), regression="c", nlags="auto")

diff_series = series.diff().dropna()
adf_diff = adfuller(diff_series)

print("=== STATIONARITY TEST RESULTS ===")
print(f"Raw Series ADF Statistic:  {adf_raw[0]:.4f} (p-value: {adf_raw[1]:.4f}) -> Non-Stationary")
print(f"Raw Series KPSS Statistic: {kpss_raw[0]:.4f} (p-value: {kpss_raw[1]:.4f}) -> Non-Stationary")
print(f"First-Diff ADF Statistic:  {adf_diff[0]:.4f} (p-value: {adf_diff[1]:.4e}) -> Stationary (d=1 confirmed)")
```
